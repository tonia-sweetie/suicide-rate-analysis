### Q1: Which continent has the highest average suicide rate?

```python
continent_average = suicide_info.groupby(['Continent', 'Year']).mean().sort_values(by = 'Suicide Rate')
```

```python
continent_average.plot(kind = 'bar', title = 'Average Suicide Rate by Continent', figsize = (16, 4))

plt.show()

print('Insight: Europe has the highest average suicide rate, followed by Oceania, South America, Asia and Africa interchanging over the years with North America having the lowest average suicide rate.')
```

```python
country_avg = suicide_info.groupby('Country Name')['Suicide Rate'].mean().reset_index()
country_avg.rename(columns={'Suicide Rate': 'Avg_Suicide_Rate'}, inplace=True)

continent_info = suicide_info[['Country Name', 'Continent']].drop_duplicates()
country_avg = country_avg.merge(continent_info, on='Country Name', how='left')

# Step 2: Compute global average
global_avg = country_avg['Avg_Suicide_Rate'].mean()

# Step 3: Flag countries above global average
country_avg['Above_Global_Avg'] = country_avg['Avg_Suicide_Rate'] > global_avg

# Step 4: Calculate percentage per continent
continent_summary = country_avg.groupby('Continent').agg(
    Total_Countries=('Country Name', 'count'),
    Above_Avg_Countries=('Above_Global_Avg', 'sum')
).reset_index()

continent_summary['Percentage_Above_Avg'] = round(
    (continent_summary['Above_Avg_Countries'] / continent_summary['Total_Countries']) * 100, 2
)

print(continent_summary[['Continent', 'Percentage_Above_Avg']])
```
